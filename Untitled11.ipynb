CELL 6 — Encode diagnosis

We will use:

B = 0 → Benign

M = 1 → Malignant

```python
df["diagnosis_encoded"] = df["diagnosis"].map({
    "B": 0,
    "M": 1
})

print(df[["diagnosis", "diagnosis_encoded"]].head())
```

FIGURE 1 — Diagnosis Distribution

```python
plt.figure(figsize=(7, 5))

ax = sns.countplot(
    data=df,
    x="diagnosis"
)

plt.title(
    "Breast Cancer Diagnosis Distribution",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Diagnosis")
plt.ylabel("Number of Patients")

for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()

plt.savefig(
    "Fig_1_Diagnosis_Distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

CELL 7 — Prepare X and y

```python
X = df.drop(
    columns=[
        "diagnosis",
        "diagnosis_encoded"
    ]
)

y = df["diagnosis_encoded"]

print("Number of features:", X.shape[1])
print("Number of samples:", X.shape[0])
```

FIGURE 2 — Correlation Heatmap

```python
plt.figure(figsize=(15, 12))

corr = X.corr()

sns.heatmap(
    corr,
    cmap="coolwarm",
    center=0,
    linewidths=0.2
)

plt.title(
    "Pearson Correlation Heatmap of Breast Cancer Features",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    "Fig_2_Correlation_Heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

CELL 8 — Train/Test Split

We use an 80/20 stratified split.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

print("\nTraining distribution:")
print(y_train.value_counts())

print("\nTesting distribution:")
print(y_test.value_counts())
```

CELL 9 — 5-Fold Cross Validation

```python
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

CELL 10 — Create the four models

We'll use:

Logistic Regression

Random Forest

XGBoost

MLP

```python
# ============================================================
# LOGISTIC REGRESSION
# ============================================================

lr_pipeline = ImbPipeline([
    ("scaler", StandardScaler()),
    ("smote", SMOTE(random_state=42)),
    ("model", LogisticRegression(
        max_iter=3000,
        random_state=42
    ))
])


# ============================================================
# RANDOM FOREST
# ============================================================

rf_pipeline = ImbPipeline([
    ("smote", SMOTE(random_state=42)),
    ("model", RandomForestClassifier(
        random_state=42,
        n_jobs=-1
    ))
])


# ============================================================
# XGBOOST
# ============================================================

xgb_pipeline = ImbPipeline([
    ("smote", SMOTE(random_state=42)),
    ("model", XGBClassifier(
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    ))
])


# ============================================================
# MLP
# ============================================================

mlp_pipeline = ImbPipeline([
    ("scaler", StandardScaler()),
    ("smote", SMOTE(random_state=42)),
    ("model", MLPClassifier(
        max_iter=1500,
        early_stopping=True,
        random_state=42
    ))
])

print("All four models created.")
```

CELL 11 — Hyperparameter tuning

```python
lr_params = {
    "model__C": [0.1, 1, 10],
    "model__class_weight": [None, "balanced"]
}

rf_params = {
    "model__n_estimators": [200, 300],
    "model__max_depth": [None, 10, 20],
    "model__min_samples_split": [2, 5]
}

xgb_params = {
    "model__n_estimators": [200, 300],
    "model__max_depth": [3, 5],
    "model__learning_rate": [0.03, 0.05],
    "model__subsample": [0.8, 1.0]
}

mlp_params = {
    "model__hidden_layer_sizes": [
        (32,),
        (64, 32),
        (128, 64)
    ],
    "model__activation": ["relu"],
    "model__alpha": [0.0001, 0.001]
}
```

CELL 12 — Run GridSearchCV

```python
print("Tuning Logistic Regression...")

lr_search = GridSearchCV(
    lr_pipeline,
    lr_params,
    scoring="f1",
    cv=cv,
    n_jobs=-1,
    verbose=1
)

lr_search.fit(X_train, y_train)


print("\nTuning Random Forest...")

rf_search = GridSearchCV(
    rf_pipeline,
    rf_params,
    scoring="f1",
    cv=cv,
    n_jobs=-1,
    verbose=1
)

rf_search.fit(X_train, y_train)


print("\nTuning XGBoost...")

xgb_search = GridSearchCV(
    xgb_pipeline,
    xgb_params,
    scoring="f1",
    cv=cv,
    n_jobs=-1,
    verbose=1
)

xgb_search.fit(X_train, y_train)


print("\nTuning MLP...")

mlp_search = GridSearchCV(
    mlp_pipeline,
    mlp_params,
    scoring="f1",
    cv=cv,
    n_jobs=-1,
    verbose=1
)

mlp_search.fit(X_train, y_train)

print("\nTUNING COMPLETED!")
```

CELL 13 — Show best parameters

```python
searches = {
    "Logistic Regression": lr_search,
    "Random Forest": rf_search,
    "XGBoost": xgb_search,
    "MLP Neural Network": mlp_search
}

for name, search in searches.items():

    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    print("Best Parameters:")
    print(search.best_params_)

    print(
        "Best CV F1:",
        round(search.best_score_, 4)
    )
```

CELL 14 — Store optimized models

```python
best_models = {
    "Logistic Regression": lr_search.best_estimator_,
    "Random Forest": rf_search.best_estimator_,
    "XGBoost": xgb_search.best_estimator_,
    "MLP Neural Network": mlp_search.best_estimator_
}
```

CELL 15 — Generate predictions

```python
predictions = {}
probabilities = {}

for name, model in best_models.items():

    predictions[name] = model.predict(X_test)

    probabilities[name] = (
        model.predict_proba(X_test)[:, 1]
    )

print("Predictions generated successfully.")
```

CELL 16 — Calculate performance

```python
results = []

for name in best_models.keys():

    pred = predictions[name]
    prob = probabilities[name]

    results.append({

        "Model": name,

        "Accuracy": accuracy_score(
            y_test,
            pred
        ),

        "Precision": precision_score(
            y_test,
            pred
        ),

        "Recall": recall_score(
            y_test,
            pred
        ),

        "F1-Score": f1_score(
            y_test,
            pred
        ),

        "ROC-AUC": roc_auc_score(
            y_test,
            prob
        )
    })

performance_df = pd.DataFrame(results)

display(
    performance_df.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1-Score": "{:.4f}",
        "ROC-AUC": "{:.4f}"
    })
)
```

6. Weighted Hybrid Ensemble Comparison

```python
# ============================================================
# WEIGHTED HYBRID ENSEMBLE
# ============================================================

cv_scores = {
    name: search.best_score_
    for name, search in searches.items()
}

weights = np.array(
    list(cv_scores.values())
)

weights = weights / weights.sum()

model_names = list(
    cv_scores.keys()
)

ensemble_weights = dict(
    zip(
        model_names,
        weights
    )
)

print("Ensemble weights:")

for name, weight in ensemble_weights.items():

    print(
        f"{name}: {weight:.4f}"
    )
```

```python
ensemble_probability = np.zeros(
    len(X_test)
)

for name in model_names:

    ensemble_probability += (
        probabilities[name]
        *
        ensemble_weights[name]
    )

ensemble_prediction = (
    ensemble_probability >= 0.50
).astype(int)
```

```python
ensemble_results = {

    "Model":
        "Weighted Hybrid Ensemble",

    "Accuracy":
        accuracy_score(
            y_test,
            ensemble_prediction
        ),

    "Precision":
        precision_score(
            y_test,
            ensemble_prediction
        ),

    "Recall":
        recall_score(
            y_test,
            ensemble_prediction
        ),

    "F1-Score":
        f1_score(
            y_test,
            ensemble_prediction
        ),

    "ROC-AUC":
        roc_auc_score(
            y_test,
            ensemble_probability
        )
}

ensemble_df = pd.DataFrame([
    ensemble_results
])

display(ensemble_df)
```

```python
final_results = pd.concat(
    [
        performance_df[
            [
                "Model",
                "Accuracy",
                "Precision",
                "Recall",
                "F1-Score",
                "ROC-AUC"
            ]
        ],
        ensemble_df
    ],
    ignore_index=True
)

display(final_results)
```

```python
# ============================================================
# ADDITIONAL FIGURE — ENSEMBLE COMPARISON
# ============================================================

metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1-Score",
    "ROC-AUC"
]

ensemble_plot = final_results.set_index(
    "Model"
)[metrics]

ensemble_plot.plot(
    kind="bar",
    figsize=(13, 7)
)

plt.title(
    "Performance Comparison Including Weighted Hybrid Ensemble",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Model")
plt.ylabel("Score")

plt.ylim(
    0,
    1.05
)

plt.xticks(
    rotation=20
)

plt.legend(
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.tight_layout()

plt.savefig(
    "Additional_Fig_Weighted_Ensemble_Comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# STATISTICAL SIGNIFICANCE — McNEMAR TEST
# ============================================================

from statsmodels.stats.contingency_tables import mcnemar

best_model_name = performance_df.loc[
    performance_df["F1-Score"].idxmax(),
    "Model"
]

best_pred = predictions[
    best_model_name
]

statistical_results = []

for name in best_models.keys():

    if name == best_model_name:
        continue

    other_pred = predictions[name]

    best_correct = (
        best_pred == y_test.values
    )

    other_correct = (
        other_pred == y_test.values
    )

    table = np.array([
        [
            np.sum(
                best_correct &
                other_correct
            ),
            np.sum(
                best_correct &
                ~other_correct
            )
        ],
        [
            np.sum(
                ~best_correct &
                other_correct
            ),
            np.sum(
                ~best_correct &
                ~other_correct
            )
        ]
    ])

    result = mcnemar(
        table,
        exact=True
    )

    statistical_results.append({

        "Comparison":
            f"{best_model_name} vs {name}",

        "p-value":
            result.pvalue,

        "Significant (p < 0.05)":
            result.pvalue < 0.05
    })

statistical_df = pd.DataFrame(
    statistical_results
)

display(statistical_df)
```
